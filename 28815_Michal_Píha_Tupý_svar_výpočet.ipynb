```python
# ==============================================================================
# 28815 - Bc. Michal Píha
# Téma: Pevnostní výpočet TUPÉHO SVARU - Text mimo graf
# Prostředí: Google Colab
# ==============================================================================

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

def vypocet_tupy_svar_clean(F_kN, L_mm, t_mm, alpha_deg, Re_MPa, k_bezpecnost):
    """
    Provádí výpočet tupého svaru a zobrazuje výsledky v grafu.
    """

    # 1. PŘÍPRAVA DAT
    F_N = F_kN * 1000
    alpha_rad = np.radians(alpha_deg)

    # Rozklad síly
    Fn = F_N * np.cos(alpha_rad)
    Ft = F_N * np.sin(alpha_rad)

    # Plocha průřezu
    S = L_mm * t_mm

    # 2. VÝPOČET PRO ZVOLENÝ BOD
    if S > 0:
        sigma = Fn / S
        tau = Ft / S
        sigma_red = np.sqrt(sigma**2 + 3 * tau**2)
    else:
        sigma = 0; tau = 0; sigma_red = 0

    # 3. DOVOLENÉ NAPĚTÍ A POSOUZENÍ
    sigma_dov = Re_MPa / k_bezpecnost
    status = "VYHOVUJE" if sigma_red <= sigma_dov else "NEVYHOVUJE"
    color_status = "green" if status == "VYHOVUJE" else "red"

    # 4. VIZUALIZACE A VÝPOČET KŘIVEK
    fig, ax = plt.subplots(figsize=(14, 6))

    # Upravíme okraje
    plt.subplots_adjust(right=0.75)

    # Rozsah tloušťky
    t_range = np.linspace(1, 50, 150)
    S_range = L_mm * t_range

    # Křivky
    sigma_curve = Fn / S_range
    tau_curve = Ft / S_range
    sigma_red_curve = np.sqrt(sigma_curve**2 + 3 * tau_curve**2)

    # Vykreslení
    ax.plot(t_range, sigma_curve, label=r'Normálové napětí $\sigma$', color='blue', linestyle=':', alpha=0.6)
    ax.plot(t_range, tau_curve, label=r'Smykové napětí $\tau$', color='green', linestyle=':', alpha=0.6)
    ax.plot(t_range, sigma_red_curve, label=r'Redukované napětí $\sigma_{red}$', color='red', linewidth=2.5)

    # Limit
    ax.axhline(y=sigma_dov, color='black', linestyle='--', linewidth=1.5, label=f'Dovolené napětí ({sigma_dov:.1f} MPa)')

    # Zvýraznění bodu
    ax.scatter([t_mm], [sigma_red], color=color_status, s=200, zorder=5, edgecolors='black')
    ax.axvline(x=t_mm, color='gray', linestyle='-.', alpha=0.4)

    # 5. INFORMAČNÍ BOX - UMÍSTĚNÍ MIMO GRAF
    text_vysledky = (
        f"VSTUPNÍ PARAMETRY:\n"
        f"{'-'*28}\n"
        f"Síla F      = {F_kN} kN\n"
        f"Úhel síly   = {alpha_deg}°\n"
        f"Délka L     = {L_mm} mm\n"
        f"Tloušťka t  = {t_mm} mm\n\n"
        f"VÝSLEDKY NAPĚTÍ:\n"
        f"{'-'*28}\n"
        f"Normálové σ = {sigma:.2f} MPa\n"
        f"Smykové   τ = {tau:.2f} MPa\n"
        f"Reduk. σ_red = {sigma_red:.2f} MPa\n\n"
        f"LIMITNÍ STAVY:\n"
        f"{'-'*28}\n"
        f"Mez kluzu Re = {Re_MPa} MPa\n"
        f"Bezpečnost k = {k_bezpecnost}\n"
        f"Dovolené σ   = {sigma_dov:.2f} MPa\n\n"
        f"POSOUZENÍ:\n"
        f"[{status}]"
    )

    # Souřadnice x=1.05 znamená "kousek vpravo od pravého okraje grafu"
    # Souřadnice y=1.00 znamená "zarovnáno s horním okrajem grafu"
    props = dict(boxstyle='round', facecolor='white', alpha=1.0, edgecolor=color_status, linewidth=2)

    ax.text(1.05, 1.00, text_vysledky, transform=ax.transAxes, fontsize=11,
            verticalalignment='top', horizontalalignment='left', bbox=props, fontfamily='monospace')

    # Popisky
    ax.set_title(f'Pevnostní analýza tupého svaru', fontsize=16, fontweight='bold')
    ax.set_xlabel('Tloušťka materiálu t [mm]', fontsize=12)
    ax.set_ylabel('Napětí [MPa]', fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.5)
    ax.legend(loc='upper right')

    # Osa Y
    top_limit = max(sigma_red, sigma_dov) * 1.4
    if top_limit < 10: top_limit = 10
    ax.set_ylim(0, top_limit)

    plt.show()

# SPUŠTĚNÍ
print("Interaktivní návrh (Výsledky vpravo):")
interact(vypocet_tupy_svar_clean,
         F_kN=FloatSlider(value=80, min=1, max=300, step=1, description='Síla F [kN]'),
         L_mm=IntSlider(value=150, min=20, max=500, step=10, description='Délka L [mm]'),
         t_mm=FloatSlider(value=15, min=2, max=50, step=0.5, description='Tloušťka t [mm]'),
         alpha_deg=IntSlider(value=30, min=0, max=90, step=5, description='Úhel síly [°]'),
         Re_MPa=IntSlider(value=235, min=150, max=500, step=5, description='Mez kluzu Re'),
         k_bezpecnost=FloatSlider(value=1.5, min=1.1, max=4.0, step=0.1, description='Bezpečnost k')
);
```
